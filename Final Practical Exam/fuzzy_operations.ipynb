# Union of Two Fuzzy Sets

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def traingular_mf(x,a,b,c):
    mu = []
    for value in x:
        if value<=a:
            mu.append(0)
        elif value>a and value<=b:
            mu.append((value-a)/(b-a))
        elif value>b and value<=c:
            mu.append((c-value)/(c-b))
        else:
            mu.append(0)
    return np.array(mu)
```

```python
x = np.linspace(0,10,1000)
set1 = traingular_mf(x,1,3,5)
set2 = traingular_mf(x,2,4,6)
union = np.maximum(set1,set2)
plt.plot(x,set1)
plt.plot(x,set2)
plt.plot(x,union)
plt.xlabel("Universe of Discourse")
plt.ylabel("Membership Value")
plt.title("Union")
plt.fill_between(x,union,color="magenta",alpha=0.2)
plt.grid(True)
plt.show()
```

# Intersection

```python
def trapezoidal_mf(x,a,b,c,d):
    mu = []
    for value in x:
        if value<=a:
            mu.append(0)
        elif value>a and value<=b:
            mu.append((value-a)/(b-a))
        elif value>b and value<=c:
            mu.append(1)
        elif value>c and value<=d:
            mu.append((d-value)/(d-c))
        else:
            mu.append(0)
    return np.array(mu)
```

```python
x = np.linspace(0,10,1000)
set1 = trapezoidal_mf(x,1,3,5,7)
set2 = trapezoidal_mf(x,2,4,6,8)
intersection = np.minimum(set1,set2)
plt.plot(x,set1)
plt.plot(x,set2)
plt.plot(x,intersection)
plt.xlabel("Universe of Discourse")
plt.ylabel("Membership Values")
plt.title("Intersection")
plt.fill_between(x,intersection,color="magenta",alpha=0.2)
plt.grid(True)
plt.show()
```

# Complement

```python
complement = 1-set1
plt.plot(x,set1)
plt.plot(x,complement)
plt.xlabel("Universe of Discourse")
plt.ylabel("Membership Values")
plt.title("Complement")
plt.fill_between(x,complement,color="magenta",alpha=0.2)
plt.grid(True)
plt.show()
```

# Algebric Sum

```python
set1 = traingular_mf(x,1,3,5)
set2 = traingular_mf(x,2,4,6)
sum = set1+set2-(set1*set2)
plt.plot(x,set1)
plt.plot(x,set2)
plt.plot(x,sum)
plt.xlabel("Universe of Discourse")
plt.ylabel("Membership Values")
plt.title("Algebric Sum")
plt.fill_between(x,sum,color="magenta",alpha=0.2)
plt.grid(True)
plt.show()
```

# Algebric Difference

```python
set1 = traingular_mf(x,1,3,5)
set2 = traingular_mf(x,2,4,6)
difference = np.minimum(set1,(1-set2))
plt.plot(x,set1)
plt.plot(x,set2)
plt.plot(x,difference)
plt.xlabel("Universe of Discourse")
plt.ylabel("Membership Values")
plt.title("Algebric Difference")
plt.fill_between(x,difference,color="magenta",alpha=0.2)
plt.grid(True)
plt.show()
```

# Cartesian Product

```python
R = np.array([[0.5,0.3,0.7],
             [0.9,0.2,0.4]])
S = np.array([[0.4,0.5],
             [0.2,0.8],
             [0.9,0.6]])
result = []
for i in range(len(R)):
    for j in range(len(R[0])):
        row = []
        for k in range(len(S)):
            for l in range(len(S[0])):
                value = np.minimum(R[i][j],S[k][l])
                row.append(value)
        result.append(row)
print(np.array(result))
```

# Max-Min Composition

```python
R = np.array([[0.5,0.3,0.7],
             [0.9,0.2,0.4]])
S = np.array([[0.4,0.5],
             [0.2,0.8],
             [0.9,0.6]])
final = []
for i in range(len(R)):
    result = []
    for j in range(len(S[0])):
        elementlist = []
        for k in range(len(S)):
            element = np.minimum(R[i][k],S[k][j])
            elementlist.append(element)
        result.append(max(elementlist))
    final.append(result)
print(np.array(final))
```

# Max-Product Composition

```python
R = np.array([[0.5,0.3,0.7],
             [0.9,0.2,0.4]])
S = np.array([[0.4,0.5],
             [0.2,0.8],
             [0.9,0.6]])
final = []
for i in range(len(R)):
    result = []
    for j in range(len(S[0])):
        elementlist = []
        for k in range(len(S)):
            element = R[i][k]*S[k][j]
            elementlist.append(element)
        result.append(max(elementlist))
    final.append(result)
print(np.array(final))
```
